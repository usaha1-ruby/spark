```python
diamond_df = spark.read.table("Practice.diamond")
diamond_df.show()
```

```python
from pyspark.sql.functions import avg

diamond_avg_price_df = diamond_df.groupBy("color").agg(avg("price")).withColumnRenamed("avg(price)", "avg_price").orderBy("color")
diamond_avg_price_df.show()
```
