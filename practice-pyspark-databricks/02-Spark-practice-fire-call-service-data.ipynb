```python
#spark is Spark session object automatically created by Databricks. Spark session is the entry point for accessing Spark programming APIs and Spark session is available through this session object.
#spark.read returns Spark DataFrameReader object reference and we can access following methods using this object
#this long form of code to create Spark dataframe is the preferred approach because this is standardize format.
 
fire_df = spark.read \
    .format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/databricks-datasets/learning-spark-v2/sf-fire/sf-fire-calls.csv")
```

```python
display(fire_df)
```

```python
fire_df.createGlobalTempView("fire_service_call_global_vw")
#[NOT_SUPPORTED_WITH_SERVERLESS] GLOBAL TEMPORARY VIEW is not supported on serverless compute. SQLSTATE: 0A000
```

```python
fire_df.createTempView("fire_service_call_vw")
```
